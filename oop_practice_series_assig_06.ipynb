# 4. **Class Variables and Class Methods**
Assignment:
Create a class Bank with a class variable bank_name. Add a class method change_bank_name(cls, name) that allows changing the bank name. Show that it affects all instances.



```python
class Bank:
    bank_name = "Default Bank"  # Class variable shared among all instances

    def __init__(self, account_holder):
        self.account_holder = account_holder  # Instance variable to store account holder's name

    @classmethod
    def change_bank_name(cls, name):
        cls.bank_name = name  # Changing the class variable using cls

    def display_account_details(self):
        print(f"Account Holder: {self.account_holder}, Bank Name: {self.bank_name}")


# Create objects for different account holders
account1 = Bank("Hussnain")
account2 = Bank("Arham")


# Display account details for both objects
account1.display_account_details()  # Output: Account Holder: Shabnam, Bank Name: Default Bank
account2.display_account_details()  # Output: Account Holder: Abeeha, Bank Name: Default Bank


# Change the bank name using class method
Bank.change_bank_name("Mezaan Bank")


# Display account details after changing the bank name
account1.display_account_details()  # Output: Account Holder: Shabnam, Bank Name: New Bank
account2.display_account_details()  # Output: Account Holder: Abeeha, Bank Name: New Bank
```

# **5. Static Variables and Static Methods**
Assignment:
Create a class MathUtils with a static method add(a, b) that returns the sum. No class or instance variables should be used.



```python
class MathUtils:

    # Static Method
    @staticmethod
    def add(a, b):
        return a + b  # This method simply adds two numbers

    # Static Variable
    call_count = 0  # This variable counts how many times the static method was called

    @staticmethod
    def count_calls():
        MathUtils.call_count += 1  # Increments call count every time a method is called
        return MathUtils.call_count

# Using Static Method
result = MathUtils.add(10, 20)
print(f"Sum: {result}")  # Output: Sum: 30

# Counting method calls
MathUtils.count_calls()
MathUtils.count_calls()
print(f"Static Method Called: {MathUtils.call_count} times")  # Output: Static Method Called: 2 times
```

# **9. Abstract Classes and Methods**
Assignment:
Use the abc module to create an abstract class Shape with an abstract method area(). Inherit a class Rectangle that implements area().

```python
from abc import ABC, abstractmethod

# Abstract base class
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass  # This method must be implemented in the child class

# Child class
class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width  # Area of rectangle

# Create object of Rectangle
r1 = Rectangle(10, 5)

# Call area method
print("Area of Rectangle:", r1.area())
```

# **15. Method Resolution Order (MRO) and Diamond Inheritance**
Assignment:

A with a method show(),

B and C that inherit from A and override show(),

D that inherits from both B and C.

Create an object of D and call show() to observe MRO.

# **16. Function Decorators**
Assignment:
Write a decorator function log_function_call that prints "Function is being called" before a function executes. Apply it to a function say_hello()

```python
# Decorator function
def simple_decorator(func):
    def wrapper():
        print("----- Start -----")   # Before the function runs
        func()                      # Original function call
        print("----- End -----")     # After the function runs
    return wrapper

# Function with decorator
@simple_decorator
def greet():
    print("Hello, Shabnam!")

# Call the decorated function
greet()
```

# **17. Class Decorators**
Assignment:
Create a class decorator add_greeting that modifies a class to add a greet() method returning "Hello from Decorator!". Apply it to a class Person.

```python
# Class decorator
def add_greeting(cls):
    # Define a new method
    def greet(self):
        return "Hello from Decorator!"

    # Attach the new method to the class
    cls.greet = greet
    return cls

# Apply the class decorator
@add_greeting
class Person:
    def __init__(self, name):
        self.name = name

# Create object and call the new method
p = Person("Samia")
print(p.name)       # Original attribute
print(p.greet())    # New method added by decorator
```

# **18. Property Decorators: @property, @setter, and @deleter**
Assignment:
Create a class Product with a private attribute _price. Use @property to get the price, @price.setter to update it, and @price.deleter to delete it.

```python
class Product:
    def __init__(self, price):
        self._price = price  # private attribute (by convention)

    @property
    def price(self):
        # Getter method
        return self._price

    @price.setter
    def price(self, value):
        # Setter method
        if value < 0:
            print("Price cannot be negative!")
        else:
            self._price = value

    @price.deleter
    def price(self):
        # Deleter method
        print("Deleting price...")
        del self._price

# Create object
p = Product(100)

# Accessing the price
print(p.price)     # Output: 100

# Setting new price
p.price = 150
print(p.price)     # Output: 150

# Trying to set a negative price
p.price = -50      # Output: Price cannot be negative!

# Deleting the price
del p.price
```

# **19. callable() and __call__()**
Assignment:
Create a class Multiplier with an __init__() to set a factor. Define a __call__() method that multiplies an input by the factor. Test it with callable() and by calling the object like a function.



```python
class Multiplier:
    def __init__(self, factor):
        self.factor = factor  # Initialize the multiplier factor

    def __call__(self, number):
        # The object can be called like a function
        return self.factor * number

# Create an object with a factor of 5
multiply_by_5 = Multiplier(5)

# Check if the object is callable
print(callable(multiply_by_5))  # Output: True

# Use the object like a function
result = multiply_by_5(10)  # It will multiply 10 by 5
print(result)  # Output: 50
```

# **20. Creating a Custom Exception**
Assignment:
Create a custom exception InvalidAgeError. Write a function check_age(age) that raises this exception if age < 18. Handle it with try...except.

```python
# Step 1: Create a custom exception
class InvalidAgeError(Exception):
    def __init__(self, message="Age cannot be less than 18"):
        self.message = message
        super().__init__(self.message)

# Step 2: Function to check age
def check_age(age):
    if age < 18:
        # Raise the custom exception if age is less than 18
        raise InvalidAgeError(f"Age {age} is not valid. Must be 18 or older.")
    else:
        # Print valid message if age is valid
        print("Age is valid")

# Step 3: Handling the exception with try...except block
try:
    check_age(15)  # This will raise the exception
except InvalidAgeError as e:
    print(e)  # Output: Age 15 is not valid. Must be 18 or older.

# Trying a valid age
try:
    check_age(20)  # This will print "Age is valid"
except InvalidAgeError as e:
    print(e)
```

# **21. Make a Custom Class Iterable**
Assignment:
Create a class Countdown that takes a start number. Implement __iter__() and __next__() to make the object iterable in a for-loop, counting down to 0.

```python
# Custom iterable class
class MyNumbers:
    def __init__(self):
        self.num = 1  # Starting number

    def __iter__(self):
        return self  # Returning the iterator object itself

    def __next__(self):
        if self.num <= 5:  # Limit to 5
            current = self.num
            self.num += 1
            return current  # Return current number and increment
        else:
            raise StopIteration  # End of iteration

# Using the iterable object in a loop
numbers = MyNumbers()

for n in numbers:
    print(n)
```
